## Reverse a List

```python
ls=[1,2,3,4,5,6]
```

```python
def reverse_lst(lst):
    return list(reversed(lst))
```

```python
print(reverse_lst(ls))
```

```python
print(ls[::-1])
```

```python
def reverse_list(lst):
    reversed_list = []
    for item in lst:
        reversed_list.insert(0, item)  # Insert each element at the beginning
    return reversed_list
```

```python
print(reverse_list(ls))
```

```python
def reverse_list_in_place(lst):
    start, end = 0, len(lst) - 1
    while start < end:
        lst[start], lst[end] = lst[end], lst[start]  # Swap
        print(start,end)
        start += 1
        end -= 1
    return lst
```

Number of Odd and Even Elements in a list
Count Even and Odd Numbers in a List

You are given a list of integers. Write a Python program that counts and returns the number of even and odd numbers in the list.

Parameters:

lst (List of integers): The list of integers where you will count the even and odd numbers.

Returns:

A tuple (even_count, odd_count) where even_count is the number of even numbers and odd_count is the number of odd numbers.

Example:

Input: lst = [1, 2, 3, 4, 5]
Output: (2, 3)

There are 2 even numbers: 2, 4

There are 3 odd numbers: 1, 3, 5

```python
def count_even_odd(lst):
    # Your code goes here
    leven=[i for i in lst if i%2==0]
    lodd=[i for i in lst if i%2!=0]
    return leven,lodd  
```

```python
lst=[1,2,3,4,5]
even,odd=count_even_odd(lst)
print(f"even number {even},odd numbers {odd}")
```

Number of Odd and Even Elements in a list
Count Even and Odd Numbers in a List

You are given a list of integers. Write a Python program that counts and returns the number of even and odd numbers in the list.

Parameters:

lst (List of integers): The list of integers where you will count the even and odd numbers.

Returns:

A tuple (even_count, odd_count) where even_count is the number of even numbers and odd_count is the number of odd numbers.

Example:

Input: lst = [1, 2, 3, 4, 5]
Output: (2, 3)

There are 2 even numbers: 2, 4

There are 3 odd numbers: 1, 3, 5

```python
def count_even_odd2(lst):
    # Your code goes here
    leven=[i for i in lst if i%2==0]
    lodd=[i for i in lst if i%2!=0]
    return len(leven),len(lodd)  
```

```python
print(count_even_odd2(lst))
```

```python
lst=[1,2,3,4,5]
even,odd=count_even_odd2(lst)
print(f"even number {even},odd numbers {odd}")
```
