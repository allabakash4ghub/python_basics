Step 1: View Documentation in Python

Inside Python shell:

```python
import math_utils
help(math_utils)
```

🔹 Step 2: Generate HTML Documentation

From the terminal (in the same folder as math_utils.py), run:

```python
pydoc -w math_utils
```

👉 This creates a file: math_utils.html

Open it in your browser, and you’ll see nicely formatted documentation with:

Module description

Function names & signatures

Function docstrings


###########

🔹 Step 3: (Optional) Generate Docs for All Modules in a Package

If you have a package, say:

my_project/
    __init__.py
    math_utils.py
    string_utils.py


You can run:

pydoc -w my_project


This generates HTML docs for the whole package automatically
