## Reverse a List

```python
ls=[1,2,3,4,5,6]
```

```python
def reverse_lst(lst):
    return list(reversed(lst))
```

```python
print(reverse_lst(ls))
```

```python
print(ls[::-1])
```

```python
lst2=[]
def reverse_lst1(lst):
    for i in range(len(lst)):
        lst2.insert(0,lst[i])
    return lst2
```

```python
print(reverse_lst1(ls))
```

```python
def reverse_list(lst):
    reversed_list = []
    for item in lst:
        reversed_list.insert(0, item)  # Insert each element at the beginning
    return reversed_list
```

```python
print(reverse_list(ls))
```
