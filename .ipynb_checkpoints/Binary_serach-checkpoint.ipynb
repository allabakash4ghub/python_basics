# O(log n) → Logarithmic Time
A classic example is Binary Search.

Idea:
You repeatedly divide the problem in half.

Each step reduces the remaining work by half → logarithmic growth.

# Why O(log n)?

Each step cuts the range in half → log₂(n) checks.

E.g. 1,000,000 elements → only ~20 comparisons.

Super fast for large data — but needs sorted data!


```python
# Suppose you have a sorted list:
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid  # Found!
        elif arr[mid] < target:
            low = mid + 1  # Search right half
        else:
            high = mid - 1  # Search left half
    return -1  # Not found

# Example
nums = [1, 3, 5, 7, 9, 11, 13]
result = binary_search(nums, 7)
print(result)  # 3 (index of 7)
```

# O(n²) → Quadratic Time
A classic example is nested loops.
For each element, you loop over every other element.
# Why O(n²)?

Outer loop runs n times.

Inner loop runs up to n times for each outer loop.

So total work: ~n * n = n².

```python
def find_duplicates_brute_force(arr):
    n = len(arr)
    duplicates = []
    for i in range(n):
        for j in range(i+1, n):
            if arr[i] == arr[j] and arr[i] not in duplicates:
                duplicates.append(arr[i])
    return duplicates

# Example
data = [1, 2, 3, 2, 1, 4, 5, 1]
print(find_duplicates_brute_force(data))  # [1, 2]
```

```python
# Another O(n²) example: Bubble Sort

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]

# Example
nums = [5, 3, 2, 1, 4]
bubble_sort(nums)
print(nums)  # [1, 2, 3, 4, 5]
```

# Summary
| Complexity   | Typical Example           | Key idea                         |
| ------------ | ------------------------- | -------------------------------- |
| **O(log n)** | Binary Search             | Divide problem in half each step |
| **O(n²)**    | Nested loops, Bubble Sort | Compare every pair               |

# Fast check:
| Input size | Binary search steps (log₂n) | Nested loops steps (n²) |
| ---------- | --------------------------- | ----------------------- |
| 10         | \~3                         | 100                     |
| 100        | \~7                         | 10,000                  |
| 1,000      | \~10                        | 1,000,000               |
