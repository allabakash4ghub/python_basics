````python
# Python OOP Concepts - Complete Guide with Examples

## 1. Classes and Objects

A **class** is a blueprint for creating objects. An **object** is an instance of a class.

```python
class Car:
    # Class attribute (shared by all instances)
    wheels = 4
    
    # Constructor method
    def __init__(self, brand, model, year):
        # Instance attributes (unique to each object)
        self.brand = brand
        self.model = model
        self.year = year
    
    # Instance method
    def start_engine(self):
        return f"{self.brand} {self.model} engine started!"
    
    # Instance method
    def get_info(self):
        return f"{self.year} {self.brand} {self.model}"

# Creating objects (instances)
car1 = Car("Toyota", "Camry", 2023)
car2 = Car("Honda", "Civic", 2022)

print(car1.get_info())  # Output: 2023 Toyota Camry
print(car2.start_engine())  # Output: Honda Civic engine started!
print(Car.wheels)  # Output: 4 (accessing class attribute)
```

## 2. Encapsulation

Encapsulation is the bundling of data and methods within a class, and controlling access to them using access modifiers.

```python
class BankAccount:
    def __init__(self, account_number, initial_balance):
        self.account_number = account_number  # Public attribute
        self._balance = initial_balance       # Protected attribute (convention)
        self.__pin = "1234"                  # Private attribute (name mangling)
    
    # Public method
    def get_balance(self):
        return self._balance
    
    # Public method
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            return f"Deposited ${amount}. New balance: ${self._balance}"
        return "Invalid deposit amount"
    
    # Public method
    def withdraw(self, amount, pin):
        if self.__verify_pin(pin) and amount <= self._balance:
            self._balance -= amount
            return f"Withdrew ${amount}. Remaining balance: ${self._balance}"
        return "Invalid transaction"
    
    # Private method
    def __verify_pin(self, pin):
        return pin == self.__pin
    
    # Protected method
    def _calculate_interest(self, rate):
        return self._balance * rate

# Usage
account = BankAccount("12345", 1000)
print(account.get_balance())  # Output: 1000
print(account.deposit(500))   # Output: Deposited $500. New balance: $1500

# This works (public attribute)
print(account.account_number)  # Output: 12345

# This works but shouldn't be accessed directly (protected)
print(account._balance)  # Output: 1500

# This will cause AttributeError (private attribute)
# print(account.__pin)  # AttributeError

# But you can access it using name mangling (not recommended)
print(account._BankAccount__pin)  # Output: 1234
```

## 3. Inheritance

Inheritance allows a class to inherit attributes and methods from another class.

```python
# Base class (Parent class)
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species
    
    def make_sound(self):
        return f"{self.name} makes a sound"
    
    def eat(self):
        return f"{self.name} is eating"

# Derived class (Child class)
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, "Canine")  # Call parent constructor
        self.breed = breed
    
    # Method overriding
    def make_sound(self):
        return f"{self.name} barks: Woof! Woof!"
    
    # Additional method specific to Dog
    def fetch(self):
        return f"{self.name} is fetching the ball"

class Cat(Animal):
    def __init__(self, name, color):
        super().__init__(name, "Feline")
        self.color = color
    
    # Method overriding
    def make_sound(self):
        return f"{self.name} meows: Meow!"
    
    def climb(self):
        return f"{self.name} is climbing a tree"

# Usage
dog = Dog("Buddy", "Golden Retriever")
cat = Cat("Whiskers", "Orange")

print(dog.make_sound())  # Output: Buddy barks: Woof! Woof!
print(cat.make_sound())  # Output: Whiskers meows: Meow!
print(dog.eat())         # Output: Buddy is eating (inherited method)
print(dog.fetch())       # Output: Buddy is fetching the ball
print(f"Dog breed: {dog.breed}")  # Output: Dog breed: Golden Retriever
```

## 4. Polymorphism

Polymorphism allows objects of different classes to be treated as objects of a common base class.

```python
# Base class
class Shape:
    def area(self):
        pass
    
    def perimeter(self):
        pass

# Derived classes
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius ** 2
    
    def perimeter(self):
        return 2 * 3.14159 * self.radius

class Triangle(Shape):
    def __init__(self, base, height, side1, side2, side3):
        self.base = base
        self.height = height
        self.side1 = side1
        self.side2 = side2
        self.side3 = side3
    
    def area(self):
        return 0.5 * self.base * self.height
    
    def perimeter(self):
        return self.side1 + self.side2 + self.side3

# Polymorphism in action
def print_shape_info(shape):
    """This function works with any Shape object"""
    print(f"Area: {shape.area():.2f}")
    print(f"Perimeter: {shape.perimeter():.2f}")
    print("-" * 20)

# Creating different shape objects
shapes = [
    Rectangle(5, 4),
    Circle(3),
    Triangle(6, 4, 3, 4, 5)
]

# Using polymorphism - same function works with different object types
for shape in shapes:
    print(f"Shape: {shape.__class__.__name__}")
    print_shape_info(shape)

# Output will show area and perimeter for each shape
```

## 5. Abstraction

Abstraction hides complex implementation details and shows only essential features.

```python
from abc import ABC, abstractmethod

# Abstract base class
class Vehicle(ABC):
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    
    @abstractmethod
    def start_engine(self):
        """Abstract method - must be implemented by subclasses"""
        pass
    
    @abstractmethod
    def stop_engine(self):
        """Abstract method - must be implemented by subclasses"""
        pass
    
    # Concrete method (can be used by all subclasses)
    def get_info(self):
        return f"{self.brand} {self.model}"

# Concrete implementation
class ElectricCar(Vehicle):
    def __init__(self, brand, model, battery_capacity):
        super().__init__(brand, model)
        self.battery_capacity = battery_capacity
    
    def start_engine(self):
        return f"{self.brand} {self.model} electric motor started silently"
    
    def stop_engine(self):
        return f"{self.brand} {self.model} electric motor stopped"
    
    def charge_battery(self):
        return f"Charging {self.battery_capacity}kWh battery"

class GasCar(Vehicle):
    def __init__(self, brand, model, fuel_capacity):
        super().__init__(brand, model)
        self.fuel_capacity = fuel_capacity
    
    def start_engine(self):
        return f"{self.brand} {self.model} gas engine started with a roar"
    
    def stop_engine(self):
        return f"{self.brand} {self.model} gas engine stopped"
    
    def fill_tank(self):
        return f"Filling {self.fuel_capacity}L fuel tank"

# Usage
# vehicle = Vehicle("Generic", "Car")  # This would raise TypeError

electric_car = ElectricCar("Tesla", "Model 3", 75)
gas_car = GasCar("Ford", "Mustang", 60)

print(electric_car.start_engine())
print(electric_car.charge_battery())
print(gas_car.start_engine())
print(gas_car.fill_tank())
```

## 6. Special Methods (Magic Methods)

Special methods allow you to define how objects behave with built-in operations.

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    def __str__(self):
        """String representation for users"""
        return f"'{self.title}' by {self.author}"
    
    def __repr__(self):
        """String representation for developers"""
        return f"Book('{self.title}', '{self.author}', {self.pages})"
    
    def __len__(self):
        """Length of the book (number of pages)"""
        return self.pages
    
    def __eq__(self, other):
        """Equality comparison"""
        if isinstance(other, Book):
            return (self.title == other.title and 
                   self.author == other.author)
        return False
    
    def __lt__(self, other):
        """Less than comparison (by pages)"""
        if isinstance(other, Book):
            return self.pages < other.pages
        return NotImplemented
    
    def __add__(self, other):
        """Addition (combine pages)"""
        if isinstance(other, Book):
            return self.pages + other.pages
        return NotImplemented

# Usage
book1 = Book("1984", "George Orwell", 328)
book2 = Book("Animal Farm", "George Orwell", 112)
book3 = Book("1984", "George Orwell", 328)

print(str(book1))     # Output: '1984' by George Orwell
print(repr(book1))    # Output: Book('1984', 'George Orwell', 328)
print(len(book1))     # Output: 328
print(book1 == book3) # Output: True
print(book1 < book2)  # Output: False
print(book1 + book2)  # Output: 440
```

## 7. Property Decorators

Properties allow you to access methods like attributes and add validation.

```python
class Temperature:
    def __init__(self, celsius=0):
        self._celsius = celsius
    
    @property
    def celsius(self):
        """Getter for celsius"""
        return self._celsius
    
    @celsius.setter
    def celsius(self, value):
        """Setter for celsius with validation"""
        if value < -273.15:
            raise ValueError("Temperature cannot be below absolute zero")
        self._celsius = value
    
    @property
    def fahrenheit(self):
        """Convert celsius to fahrenheit"""
        return (self._celsius * 9/5) + 32
    
    @fahrenheit.setter
    def fahrenheit(self, value):
        """Set temperature using fahrenheit"""
        if value < -459.67:
            raise ValueError("Temperature cannot be below absolute zero")
        self._celsius = (value - 32) * 5/9
    
    @property
    def kelvin(self):
        """Convert celsius to kelvin"""
        return self._celsius + 273.15

# Usage
temp = Temperature(25)
print(f"Celsius: {temp.celsius}")     # Output: Celsius: 25
print(f"Fahrenheit: {temp.fahrenheit}") # Output: Fahrenheit: 77.0
print(f"Kelvin: {temp.kelvin}")       # Output: Kelvin: 298.15

# Setting temperature using fahrenheit
temp.fahrenheit = 100
print(f"Celsius: {temp.celsius}")     # Output: Celsius: 37.77777777777778

# This will raise ValueError
# temp.celsius = -300
```

## 8. Class Methods and Static Methods

```python
class Employee:
    # Class attribute
    company_name = "Tech Corp"
    total_employees = 0
    
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.total_employees += 1
    
    # Instance method
    def get_details(self):
        return f"{self.name}: ${self.salary}"
    
    # Class method - works with class, not instance
    @classmethod
    def get_total_employees(cls):
        return f"Total employees: {cls.total_employees}"
    
    @classmethod
    def from_string(cls, emp_str):
        """Alternative constructor"""
        name, salary = emp_str.split('-')
        return cls(name, int(salary))
    
    # Static method - doesn't need class or instance
    @staticmethod
    def is_workday(day):
        return day.lower() not in ['saturday', 'sunday']
    
    @staticmethod
    def calculate_tax(salary):
        if salary > 50000:
            return salary * 0.3
        return salary * 0.2

# Usage
emp1 = Employee("Alice", 60000)
emp2 = Employee("Bob", 45000)

# Instance method
print(emp1.get_details())  # Output: Alice: $60000

# Class method
print(Employee.get_total_employees())  # Output: Total employees: 2

# Alternative constructor using class method
emp3 = Employee.from_string("Charlie-55000")
print(emp3.get_details())  # Output: Charlie: $55000

# Static methods
print(Employee.is_workday("Monday"))    # Output: True
print(Employee.calculate_tax(60000))    # Output: 18000.0
```

## Key Takeaways

1. **Classes and Objects**: Classes are blueprints; objects are instances
2. **Encapsulation**: Bundle data and methods; control access with public, protected, and private attributes
3. **Inheritance**: Child classes inherit from parent classes; use `super()` to access parent methods
4. **Polymorphism**: Same interface works with different object types
5. **Abstraction**: Hide implementation details; use abstract classes for common interfaces
6. **Special Methods**: Customize object behavior with built-in operations
7. **Properties**: Use decorators to create attribute-like methods with validation
8. **Class/Static Methods**: Methods that work at class level or independently of instances

These concepts work together to create well-organized, maintainable, and reusable code in Python.
````
